## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Create copies so that the original loaded datasets remain unchanged
coverage_clean = coverage_df.copy()
incidence_clean = incidence_df.copy()
cases_clean = cases_df.copy()
introduction_clean = introduction_df.copy()
schedule_clean = schedule_df.copy()


# 1. Remove leading/trailing spaces from text columns

def clean_text_columns(df):
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()
    return df

coverage_clean = clean_text_columns(coverage_clean)
incidence_clean = clean_text_columns(incidence_clean)
cases_clean = clean_text_columns(cases_clean)
introduction_clean = clean_text_columns(introduction_clean)
schedule_clean = clean_text_columns(schedule_clean)


# 2. Convert YEAR columns to nullable integer type

coverage_clean['YEAR'] = pd.to_numeric(
    coverage_clean['YEAR'], errors='coerce'
).astype('Int64')

incidence_clean['YEAR'] = pd.to_numeric(
    incidence_clean['YEAR'], errors='coerce'
).astype('Int64')

cases_clean['YEAR'] = pd.to_numeric(
    cases_clean['YEAR'], errors='coerce'
).astype('Int64')

introduction_clean['YEAR'] = pd.to_numeric(
    introduction_clean['YEAR'], errors='coerce'
).astype('Int64')

schedule_clean['YEAR'] = pd.to_numeric(
    schedule_clean['YEAR'], errors='coerce'
).astype('Int64')


# 3. Convert important numerical variables

coverage_clean['TARGET_NUMBER'] = pd.to_numeric(
    coverage_clean['TARGET_NUMBER'], errors='coerce'
)

coverage_clean['DOSES'] = pd.to_numeric(
    coverage_clean['DOSES'], errors='coerce'
)

coverage_clean['COVERAGE'] = pd.to_numeric(
    coverage_clean['COVERAGE'], errors='coerce'
)

incidence_clean['INCIDENCE_RATE'] = pd.to_numeric(
    incidence_clean['INCIDENCE_RATE'], errors='coerce'
)

cases_clean['CASES'] = pd.to_numeric(
    cases_clean['CASES'], errors='coerce'
)

schedule_clean['SCHEDULEROUNDS'] = pd.to_numeric(
    schedule_clean['SCHEDULEROUNDS'], errors='coerce'
)


# 4. Handle invalid vaccination coverage values
# Coverage should be between 0 and 100 percent

coverage_clean.loc[
    (coverage_clean['COVERAGE'] < 0) |
    (coverage_clean['COVERAGE'] > 100),
    'COVERAGE'
] = np.nan


# 5. Handle impossible negative values

coverage_clean.loc[
    coverage_clean['TARGET_NUMBER'] < 0,
    'TARGET_NUMBER'
] = np.nan

coverage_clean.loc[
    coverage_clean['DOSES'] < 0,
    'DOSES'
] = np.nan

incidence_clean.loc[
    incidence_clean['INCIDENCE_RATE'] < 0,
    'INCIDENCE_RATE'
] = np.nan

cases_clean.loc[
    cases_clean['CASES'] < 0,
    'CASES'
] = np.nan


# 6. Verify duplicate records after cleaning

coverage_clean = coverage_clean.drop_duplicates()
incidence_clean = incidence_clean.drop_duplicates()
cases_clean = cases_clean.drop_duplicates()
introduction_clean = introduction_clean.drop_duplicates()
schedule_clean = schedule_clean.drop_duplicates()


# 7. Display cleaned dataset information

print("Data wrangling completed successfully.")

print("\nCleaned Dataset Shapes:")
print("Coverage Data:", coverage_clean.shape)
print("Incidence Rate Data:", incidence_clean.shape)
print("Reported Cases Data:", cases_clean.shape)
print("Vaccine Introduction Data:", introduction_clean.shape)
print("Vaccine Schedule Data:", schedule_clean.shape)
````

````python
# Check for invalid values after data wrangling

print("========== VALIDATION AFTER CLEANING ==========")

print("\nCoverage values outside 0-100:")
print(
    ((coverage_clean['COVERAGE'] < 0) |
     (coverage_clean['COVERAGE'] > 100)).sum()
)

print("\nNegative target numbers:")
print((coverage_clean['TARGET_NUMBER'] < 0).sum())

print("\nNegative doses:")
print((coverage_clean['DOSES'] < 0).sum())

print("\nNegative incidence rates:")
print((incidence_clean['INCIDENCE_RATE'] < 0).sum())

print("\nNegative reported cases:")
print((cases_clean['CASES'] < 0).sum())

print("\nDuplicate rows after cleaning:")
print("Coverage:", coverage_clean.duplicated().sum())
print("Incidence:", incidence_clean.duplicated().sum())
print("Cases:", cases_clean.duplicated().sum())
print("Introduction:", introduction_clean.duplicated().sum())
print("Schedule:", schedule_clean.duplicated().sum())
````

### What all manipulations have you done and insights you found?

The datasets were prepared for analysis through a series of data-wrangling and consistency checks. Separate cleaned copies were created for all five datasets so that the original datasets remained unchanged.

The main data-cleaning steps included removing leading and trailing spaces from text fields, converting year fields into an appropriate integer-compatible format, and converting important analytical variables into numeric format.

Duplicate-value analysis showed that there were no exact duplicate records in any of the five datasets. Therefore, no records were removed due to duplication.

The analysis also identified invalid numerical values. Vaccination coverage contained values above the valid percentage range of 0–100, while the doses variable contained negative values. Since vaccination coverage represents the percentage of the target population vaccinated, values outside the 0–100 range were treated as missing. Negative values in target population, doses, incidence rate, and reported cases were also treated as invalid and converted to missing values.

A significant amount of missing data was observed in several variables, particularly target population, doses, and coverage in the Coverage Data. Missing values were not blindly replaced with averages or zeros because doing so could introduce artificial information into the analysis. Instead, missing values were retained and will be handled according to the requirements of individual analyses.

After the cleaning process, the datasets are more suitable for exploratory analysis, visualization, SQL storage, and Power BI reporting. The cleaned data can be used to investigate vaccination coverage trends, disease incidence, reported cases, vaccine introduction patterns, and vaccination schedules.

#### Chart - 1

````python
# Chart 1: Vaccination Coverage Trend Over Time

# Use valid coverage values only
chart1_data = coverage_clean[
    coverage_clean['COVERAGE'].notna()
].copy()

# Calculate median vaccination coverage for each year
coverage_trend = (
    chart1_data
    .groupby('YEAR')['COVERAGE']
    .median()
    .reset_index()
)

# Plot
plt.figure(figsize=(12, 6))
plt.plot(
    coverage_trend['YEAR'],
    coverage_trend['COVERAGE'],
    marker='o'
)
plt.title('Vaccination Coverage Trend Over Time')
plt.xlabel('Year')
plt.ylabel('Median Vaccination Coverage (%)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A line chart was selected to visualize the trend in median vaccination coverage over time. Since year is a time-based variable, a line chart makes it easy to observe increases, decreases, and periods of relatively stable vaccination coverage.

The chart shows a strong increase in median vaccination coverage from approximately 41% in 1980 to around 84% by 1990. Coverage continued to increase gradually during the 1990s and reached approximately 90% or higher during the 2000s.

From the mid-2000s to the late 2010s, vaccination coverage remained relatively stable at around 90–92%. A noticeable decline can be observed around 2020–2021, when median coverage dropped to approximately 86–87%. Coverage then showed a small recovery in the following years.

Overall, the data indicates substantial improvement in vaccination coverage over the long term, followed by a period of relatively stable high coverage and a recent decline.

#### Chart - 3

````python
# Chart 3: Vaccination Coverage vs Reported Measles Cases

# Select official measles vaccination coverage
measles_coverage = coverage_df[
    (coverage_df["COVERAGE_CATEGORY"] == "OFFICIAL") &
    (coverage_df["ANTIGEN_DESCRIPTION"].str.contains("Measles", case=False, na=False))
][["CODE", "NAME", "YEAR", "COVERAGE"]].copy()

# Select reported measles cases
measles_cases = cases_df[
    cases_df["DISEASE_DESCRIPTION"].str.contains("Measles", case=False, na=False)
][["CODE", "YEAR", "CASES"]].copy()

# Remove missing values
measles_coverage = measles_coverage.dropna(subset=["COVERAGE"])
measles_cases = measles_cases.dropna(subset=["CASES"])

# Merge coverage and reported cases using country and year
measles_analysis = pd.merge(
    measles_coverage,
    measles_cases,
    on=["CODE", "YEAR"],
    how="inner"
)

# Create scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=measles_analysis,
    x="COVERAGE",
    y="CASES",
    alpha=0.5
)

plt.title("Vaccination Coverage vs Reported Measles Cases")
plt.xlabel("Measles Vaccination Coverage (%)")
plt.ylabel("Reported Measles Cases")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A scatter plot was selected to analyze the relationship between measles vaccination coverage and the number of reported measles cases. It helps identify patterns, trends, clusters, and unusual observations between the two numerical variables.

##### 2. What is/are the insight(s) found from the chart?

The chart shows that most observations have relatively low numbers of reported measles cases, but a few observations have very high case counts. There is no clear strong linear relationship visible between vaccination coverage and reported cases. Most observations are concentrated at lower case counts across a wide range of vaccination coverage values. Some observations also show vaccination coverage above 100%, which may be associated with differences in reporting or estimation methods.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

However, the chart alone does not establish that vaccination coverage directly caused the differences in reported cases.

````python
queries = {
    "Coverage": "SELECT COUNT(*) FROM coverage",
    "Incidence": "SELECT COUNT(*) FROM incidence_rate",
    "Reported Cases": "SELECT COUNT(*) FROM reported_cases",
    "Vaccine Introduction": "SELECT COUNT(*) FROM vaccine_introduction",
    "Vaccine Schedule": "SELECT COUNT(*) FROM vaccine_schedule"
}
for name, query in queries.items():
    cursor.execute(query)
    count = cursor.fetchone()[0]
    print(f"{name}: {count:,} records")
````

````python
query = """
SELECT 
    YEAR,
    ROUND(AVG(COVERAGE), 2) AS average_coverage
FROM coverage
WHERE COVERAGE IS NOT NULL
GROUP BY YEAR
ORDER BY YEAR;
"""

cursor.execute(query)

results = cursor.fetchall()

print("Average Vaccination Coverage by Year:\n")

for year, coverage in results:
    print(f"{year}: {coverage}%")
````

### SQL Insight – Vaccination Coverage Trend

The SQL analysis shows a substantial increase in average vaccination coverage from 44.25% in 1980 to more than 80% during the 2000s. Coverage reached 83.71% in 2009. After 2010, average coverage remained relatively stable at approximately 78–80%, with a decline observed during 2020–2021. Coverage recovered to 78.68% by 2023.

````python
query = """
SELECT
    NAME AS country,
    ROUND(AVG(COVERAGE), 2) AS average_coverage
FROM coverage
WHERE COVERAGE IS NOT NULL
GROUP BY NAME
HAVING COUNT(COVERAGE) >= 5
ORDER BY average_coverage DESC
LIMIT 10;
"""

cursor.execute(query)

results = cursor.fetchall()

print("Top 10 Countries by Average Vaccination Coverage:\n")

for country, coverage in results:
    print(f"{country}: {coverage}%")
````

### SQL Insight – Countries with Highest Average Vaccination Coverage

The SQL analysis identifies the countries with the highest average recorded vaccination coverage. Monaco has the highest average at 96.46%, followed by Hungary at 96.10% and Turkmenistan at 95.49%. The remaining countries in the top 10 also have average coverage above 93%.

These results help identify countries with consistently high recorded vaccination coverage and can be used for comparison with countries having lower coverage.

````python
query = """
SELECT
    NAME AS country,
    ROUND(AVG(COVERAGE), 2) AS average_coverage
FROM coverage
WHERE `GROUP` = 'COUNTRIES'
  AND COVERAGE IS NOT NULL
GROUP BY NAME
HAVING COUNT(COVERAGE) >= 5
ORDER BY average_coverage ASC
LIMIT 10;
"""

cursor.execute(query)

results = cursor.fetchall()

print("10 Countries with Lowest Average Vaccination Coverage:\n")

for country, coverage in results:
    print(f"{country}: {coverage}%")
````

Papua New Guinea, Chad, Equatorial Guinea, South Sudan, Afghanistan, Angola, and Ethiopia are also among the countries with lower average recorded coverage.

These results can help identify locations where vaccination coverage data indicates potential gaps requiring further investigation and resource planning.

````python
query = """
SELECT
    DISEASE,
    MAX(DISEASE_DESCRIPTION) AS disease_description,
    SUM(CASES) AS total_cases
FROM reported_cases
WHERE CASES IS NOT NULL
GROUP BY DISEASE
ORDER BY total_cases DESC
LIMIT 10;
"""

cursor.execute(query)
results = cursor.fetchall()
print("Top 10 Diseases by Total Reported Cases:\n")
for disease, description, cases in results:
    print(f"{disease} | {description} | {cases:,.0f} cases")
````

### SQL Insight – Reported Cases by Disease

The SQL analysis shows that measles has the largest total number of reported cases in the dataset, with 144,065,673 cases. Pertussis follows with 53,089,864 cases, followed by mumps with 36,326,936 cases and typhoid with 31,153,098 cases. Rubella also accounts for a substantial number of reported cases.

These results indicate which diseases account for the largest recorded case totals in the dataset and can help guide further analysis of vaccination coverage and disease incidence.

````python
query = """
SELECT
    YEAR,
    SUM(CASES) AS total_cases
FROM reported_cases
WHERE CASES IS NOT NULL
GROUP BY YEAR
ORDER BY YEAR;
"""

cursor.execute(query)
results = cursor.fetchall()

print("Total Reported Cases by Year:\n")

for year, cases in results:
    print(f"{year}: {cases:,.0f} cases")
````

### SQL Insight – Reported Cases Trend

The SQL analysis shows that total reported cases were highest in the early years of the dataset, with 18,400,813 cases recorded in 1981. Reported cases generally declined from the 1980s through the 1990s, reaching 2,410,565 cases in 2015. After 2015, the number of reported cases increased, reaching 17,667,763 cases in 2023.

The results show considerable variation in reported cases across the years. These trends can be further investigated alongside vaccination coverage and disease-specific incidence rates.

````python
query = """
SELECT
    DISEASE,
    MAX(DISEASE_DESCRIPTION) AS disease_description,
    ROUND(AVG(INCIDENCE_RATE), 2) AS average_incidence_rate
FROM incidence_rate
WHERE INCIDENCE_RATE IS NOT NULL
GROUP BY DISEASE
ORDER BY average_incidence_rate DESC
LIMIT 10;
"""

cursor.execute(query)
results = cursor.fetchall()

print("Top 10 Diseases by Average Incidence Rate:\n")

for disease, description, rate in results:
    print(f"{disease} | {description} | {rate} average incidence rate")
````

### SQL Insight – Average Incidence Rate by Disease

The SQL analysis shows that measles has the highest average recorded incidence rate at 535.64, followed by typhoid at 348.50 and mumps at 168.05. Pertussis has an average incidence rate of 130.55, while rubella has an average of 54.35.

The remaining diseases in the top 10 have considerably lower average recorded incidence rates. These results help identify diseases with higher recorded incidence rates in the dataset and provide another perspective for comparison with reported cases and vaccination coverage.

````python
query = """
SELECT
    NAME AS country,
    SUM(CASES) AS total_cases
FROM reported_cases
WHERE `GROUP` = 'COUNTRIES'
  AND CASES IS NOT NULL
GROUP BY NAME
ORDER BY total_cases DESC
LIMIT 10;
"""

cursor.execute(query)
results = cursor.fetchall()
print("Top 10 Countries by Total Reported Cases:\n")

for country, cases in results:
    print(f"{country}: {cases:,.0f} cases")
````

### SQL Insight – Countries with Highest Total Reported Cases

The SQL analysis shows that China has the highest total number of recorded cases in the dataset, with 15,411,329 cases, followed by India with 8,591,314 cases. Nigeria and the Democratic Republic of the Congo recorded 4,220,814 and 4,172,139 cases respectively.

Other countries appearing among the top 10 include the Russian Federation, Angola, Kenya, Japan, Spain, and Ghana. These results show substantial differences in the total number of recorded cases across countries and can be used for further comparison with vaccination coverage and disease incidence.

````python
query = """
SELECT
    NAME AS country,
    ROUND(AVG(INCIDENCE_RATE), 2) AS average_incidence_rate
FROM incidence_rate
WHERE `GROUP` = 'COUNTRIES'
  AND INCIDENCE_RATE IS NOT NULL
GROUP BY NAME
HAVING COUNT(INCIDENCE_RATE) >= 5
ORDER BY average_incidence_rate DESC
LIMIT 10;
"""

cursor.execute(query)

results = cursor.fetchall()

print("Top 10 Countries by Average Incidence Rate:\n")

for country, rate in results:
    print(f"{country}: {rate} average incidence rate")
````

### SQL Insight – Countries with Highest Average Incidence Rate

The SQL analysis shows that Malawi has the highest average recorded incidence rate among the countries in the result, at 744.82, followed by Kiribati at 679.29 and Tuvalu at 662.38. Angola and Kenya also have relatively high average recorded incidence rates of 527.91 and 522.15 respectively.

Other countries appearing in the top 10 include Cook Islands, Burundi, Zambia, Madagascar, and Wallis and Futuna. These results help identify countries with higher recorded incidence rates and can be compared with reported cases and vaccination coverage to understand differences across locations.

````python
query = """
SELECT
    YEAR,
    COUNT(*) AS introductions
FROM vaccine_introduction
GROUP BY YEAR
ORDER BY YEAR;
"""

cursor.execute(query)

results = cursor.fetchall()

print("Vaccine Introductions by Year:\n")

for year, introductions in results:
    print(f"{year}: {introductions} introductions")
````

### SQL Insight – Vaccine Introductions by Year

The SQL analysis shows that the number of recorded vaccine introduction records generally increased over the period from 1940 to 2023. The dataset contains 164 introduction records per year during the early years, with the number increasing over time and reaching 4,074 records in 2019–2023.

The increase in recorded introduction records reflects the expansion of vaccine introduction information across countries and vaccine types in the dataset. One record has a missing year and is therefore listed as `None`.

````python
query = """
SELECT
    VACCINECODE,
    MAX(VACCINE_DESCRIPTION) AS vaccine_description,
    COUNT(*) AS schedule_records
FROM vaccine_schedule
GROUP BY VACCINECODE
ORDER BY schedule_records DESC
LIMIT 10;
"""

cursor.execute(query)
results = cursor.fetchall()

print("Top 10 Vaccines by Schedule Records:\n")

for code, description, records in results:
    print(f"{code} | {description} | {records} schedule records")
````

### SQL Insight – Most Common Vaccines in the Schedule

The SQL analysis shows that Td vaccine has the highest number of schedule records, with 967 records. OPV follows with 517 records, while PCV-13 has 506 records. Adult seasonal influenza vaccine has 500 schedule records.

Other frequently occurring vaccines include DTwP-Hib-HepB, adult Hepatitis B, MMR, IPV, Vitamin A supplements, and pediatric Hepatitis B. These results provide an overview of the vaccines most frequently represented in the vaccination schedule dataset.
